### 問題14 特定の値を抽出
次の配列データの中から、先頭から2つ目の値と最後の値をそれぞれ出力しましょう。  
また`2`を配列の一番最後に追加しましょう。

```python
numbers = [0, 3, 8, -4, 9, 1]
```

（目安時間：1分）

**解答**

````python
numbers = [0, 3, 8, -4, 9, 1]
print(numbers[2])
print(numbers[-1])
numbers.append(2)
print(numbers)
````

### 問題16 特定の値を削除(remove, pop)

numbers = [5, 0, 3, 8, -4, 9, 1, 5, 2]

### 問題17 フィルター
次の、偶数のみを返す関数と`filter()`を用いて先ほどの`numbers`リストから偶数のみを取り出したリストを作成しましょう。

```python
def isEven(number):
    if number % 2 == 0:
        print(f'This number, {number} is even!')
        return True
    else:
        print(f'This number, {number} is odd!')
        return False
```


（目安時間：1分）

**解答**

````python
numbers = [5, 0, 3, 8, -4, 9, 1, 5, 2]

def isEven(number):
    if number % 2 == 0:
        print(f'This number, {number} is even!')
        return True
    else:
        print(f'This number, {number} is odd!')
        return False
````

````python
list(filter(isEven, numbers))
````
