## 05.制御構文(for文)
 for文を使って繰り返しの処理を行おう
### 問題25 リストから順に表示

- リスト型の変数names(`['John', 'Kevin', 'Louis']`)から順に名前を出力しましょう。

（目安時間：1分）

**解答**

```python
names= ['John', 'Kevin', 'Louis']

for i in names:
    print(i)
```

### 問題26 range関数を使う

- range関数を用いて0~9を出力しましょう。

（目安時間：1分）

**解答**

```python
for i in range(10):
    print(i)
```

### 問題27 ループの途中で強制終了

- 問題26の処理に追加で`i=6`の時「BREAK!」と出力してループを抜け出すよう実装しましょう。

（目安時間：1分）

**解答**

```python
for i in range(10):
    if i == 6:
        print("BREAK!")
        break
        
    else:
        print(i)
```

### 問題28 ループの途中で特別な処理

- 問題26の処理に追加で`i=3`のみ「これは3です。」と出力するよう実装しましょう。

（目安時間：1分）

**解答**

```python
for i in range(10):
    if i == 3:
        continue
    else:
        print(i)
```

### 問題29 複数のリストを同時に処理

- 2つの変数(`lasts=['加藤', '佐藤', '田中']`, `firsts=['雄一', '拓也', '太郎']`)を用意
- zip関数を用いて2つのリストから姓名を同時に出力しましょう

（目安時間：2分）


**解答**

```python
lasts=['加藤', '佐藤', '田中']
firsts=['雄一', '拓也', '太郎']

for last, first in zip(lasts,firsts):
    print(last +" "+first)
```

### 問題30 リストと同時にインデックス情報の取得


- 変数(`lasts=['加藤', '佐藤', '田中']`)を用意
- enumerate関数を使って苗字とそのインデックスを取得しましょう
- 「xx番のyyです。」と出力しましょう

（目安時間：1分）


**解答**

```python
lasts=['加藤', '佐藤', '田中']
for i, last in enumerate(lasts):
    print(f"{i}番目の{last}です。")
```

### 問題31 内包表記

- 以下の2つのパターンで0から9未満となる2の倍数が入ったリスト(`[0, 2, 4, 6, 8]`)を作成しましょう。
    - 空のリスト`nums`を予め用意しforループ内で`.append()`関数を用いて要素を追加する
    - 1行で同様のリスト`numbers`を作成する
    
（目安時間：3分）


**解答**

```python
nums = []
for i in range(5):
    nums.append(i*2)
print(nums)
```

```python
numbers = [i*5 for i in range(1,10)]
numbers
```
