### 問題21 情報の取得(keys(), values(), items())
以下の辞書データから次の処理を行いましょう。

- キーのみの表示
- 値の中に`男性`という文字列が存在するか
- 全てのキーと値の表示

```python
dictionary = {
    'A': '今西',
    'B': '航平',
    'C': '男性',
    'D': '東京都',
    'E': '海外旅行',
}
```

（目安時間：3分）

**解答**

````python
dictionary = {
    'A': '今西',
    'B': '航平',
    'C': '男性',
    'D': '東京都',
    'E': '海外旅行',
}

print(dictionary.keys())
print(dictionary.values())
print(dictionary.items())
````

````python
dictionary = {
    'A': '今西',
    'B': '航平',
    'C': '男性',
    'D': '東京都',
    'E': '海外旅行',
}

for key, value in dictionary.items():
    print(f"キーは{key},バリューは、{value}")
````

````python
dictionary = {
    'A': '今西',
    'B': '航平',
    'C': '男性',
    'D': '東京都',
    'E': '海外旅行',
}

dictionary.values()
````
