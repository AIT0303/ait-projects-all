### 問題44 全ファイルの情報取得

カレントディレクトリ内の全てのファイル情報を取得しましょう。

（目安時間：3分）

**解答**

````python
import os 

for curDir, dirs, files in os.walk("."):
    for file in files:
        print(f"{curDir}/{file}")
````
